The formula for the Binomial Coeficient shows how many ways to choose  k  things out of  N:

${N\choose k}$ = $\frac{N!}{k!(N-k)!}$


```python
def factorial(n):
    total = 1
    for i in range(1, n + 1):
        total *= i
    return total

def binomial_coefficient(N, k):
    if k < 0 or k > N:
        return 0
    elif k == 0 or k == N:
        return 1
    else:
        binomial_C = factorial(N) // (factorial(N - k) * factorial(k))
        print(binomial_C)
        return binomial_C
```

```python
import numpy as np

def binomial_numpy(N, k):
    return int(np.prod(np.arange(N - k + 1, N + 1)) // np.prod(np.arange(1, k + 1)))
```

```python
if binomial_coefficient(8, 3) == binomial_numpy(8, 3):
  print("They match!")
```

This formula computes the GC content for a given sequence:

$\frac{1}{L}$$\sum_{i=1}^{L} 1[b_i\in  \{G, C\}]$

```python
def G_C_frac_counter(DNAstring):
  GC_counter = 0
  for letter in DNAstring:
    if letter == "G" or letter == "C":
      GC_counter+=1
  GC_fraction = GC_counter/len(DNAstring)
  return GC_fraction
```

```python
def G_C_frac_counter_oneline(DNAstring):
  G_C_Fraction = (DNAstring.count("G")+DNAstring.count("C"))/len(DNAstring)
  return G_C_Fraction


```
